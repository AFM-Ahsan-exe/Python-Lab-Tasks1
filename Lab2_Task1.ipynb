# ***Object Creation + Namespace (__dict__)***

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

s1 = Student("Ali", 85)

print(s1.__dict__)
```

# ***Function with Optional Arguments***

```python
def student_data(student_id, name=None, student_class=None):
    print("ID:", student_id)

    if name:
        print("Name:", name)
    if student_class:
        print("Class:", student_class)

student_data(101)
student_data(102, "Ali")
student_data(103, "Sara", "10th")
```

# ***Class Introspection***

```python
class Student:
    pass

print(type(Student))
print(Student.__dict__.keys())
print(Student.__module__)
```

# ***Class vs Object Check***

```python
class Student:
    pass

class Marks:
    pass

s = Student()
m = Marks()

print(isinstance(s, Student))
print(isinstance(m, Marks))

print(issubclass(Student, object))
print(issubclass(Marks, object))
```

# ***Modify Attributes***

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

s = Student("Ali", 80)

print("Original:", s.name, s.marks)

s.name = "Ahmad"
s.marks = 90

print("Updated:", s.name, s.marks)
```

# ***Add & Delete Attributes***

```python
class Student:
    def __init__(self, student_id, name):
        self.student_id = student_id
        self.name = name

s = Student(1, "Ali")

# Add new attribute
s.student_class = "10th"

print("Before delete:", s.__dict__)

# Delete attribute
del s.name

print("After delete:", s.__dict__)
```

# ***Display Attributes via Method***

```python
class Student:
    def __init__(self, sid, name, sclass):
        self.sid = sid
        self.name = name
        self.sclass = sclass

    def display(self):
        for key, value in self.__dict__.items():
            print(key, ":", value)

s = Student(1, "Ali", "10th")
s.display()
```

# ***Two Objects (Structured Output)***

```python
class Student:
    def __init__(self, sid, name):
        self.sid = sid
        self.name = name

    def display(self):
        print("ID:", self.sid, "| Name:", self.name)

student1 = Student(1, "Ali")
student2 = Student(2, "Sara")

student1.display()
student2.display()
```

# ***Memory Usage***

```python
import sys

class Student:
    def __init__(self, name):
        self.name = name

s = Student("Ali")

print(sys.getsizeof(s))
print(sys.getsizeof(s.__dict__))
```

# ***Reduce Memory Usage (__slots__)***

```python
import sys

class Student:
    __slots__ = ['name']

    def __init__(self, name):
        self.name = name

s = Student("Ali")

print(sys.getsizeof(s))
```

# ***1. Classes: Trapezoid & Parallelogram***

```python
class Trapezoid:
    def __init__(self, a, b, h):
        self.a = a   # parallel side 1
        self.b = b   # parallel side 2
        self.h = h   # height

    def area(self):
        return 0.5 * (self.a + self.b) * self.h


class Parallelogram:
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def area(self):
        return self.base * self.height


# Test
t = Trapezoid(4, 6, 5)
p = Parallelogram(5, 5)

print("Trapezoid Area:", t.area())
print("Parallelogram Area:", p.area())
```

# ***2. Compare Areas***

```python
def compare_areas(shape1, shape2):
    if shape1.area() > shape2.area():
        return "Shape 1 is larger"
    elif shape1.area() < shape2.area():
        return "Shape 2 is larger"
    else:
        return "Both are equal"


t = Trapezoid(4, 6, 5)
p = Parallelogram(5, 5)

print(compare_areas(t, p))
```

# ***3. Store Areas in Dictionary***

```python
t = Trapezoid(4, 6, 5)
p = Parallelogram(5, 5)

areas = {
    "Trapezoid": t.area(),
    "Parallelogram": p.area()
}

print(areas)
```

# ***4. Iterate Multiple Dimensions***

```python
trapezoid_data = [(4, 6, 5), (3, 7, 4), (5, 5, 6)]
parallelogram_data = [(5, 5), (6, 4), (3, 8)]

results = []

for t_data, p_data in zip(trapezoid_data, parallelogram_data):
    t = Trapezoid(*t_data)
    p = Parallelogram(*p_data)

    results.append({
        "Trapezoid": t.area(),
        "Parallelogram": p.area()
    })

print(results)
```

# ***5. Find Maximum Area***

```python
areas = {
    "Trapezoid": 30.0,
    "Parallelogram": 24
}

max_shape = max(areas, key=areas.get)

print("Max Area Shape:", max_shape)
print("Area:", areas[max_shape])
```
