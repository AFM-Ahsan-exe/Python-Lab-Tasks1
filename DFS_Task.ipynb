# ***Task 1: Simple DFS (Dictionary Graph)***

```python
# Graph using dictionary (adjacency list)
graph = {
    '5': ['3', '7'],
    '3': ['2', '4'],
    '7': ['8'],
    '2': [],
    '4': ['8'],
    '8': []
}

visited = set()

def dfs(node):
    if node not in visited:
        print(node, end=" ")
        visited.add(node)

        for neighbor in graph[node]:
            dfs(neighbor)

# Driver Code
print("DFS Traversal:")
dfs('5')
```

# ***ask 2: DFS Using Class (Graph with Edges)***

```python
from collections import defaultdict

class Graph:
    def __init__(self):
        self.graph = defaultdict(list)

    def add_edge(self, u, v):
        self.graph[u].append(v)

    def dfs_util(self, v, visited):
        visited[v] = True
        print(v, end=" ")

        for neighbor in self.graph[v]:
            if not visited[neighbor]:
                self.dfs_util(neighbor, visited)

    def dfs(self):
        max_node = max(self.graph.keys())
        visited = [False] * (max_node + 1)

        for i in range(max_node + 1):
            if not visited[i]:
                self.dfs_util(i, visited)


# Driver Code
g = Graph()

g.add_edge(0, 1)
g.add_edge(0, 2)
g.add_edge(1, 2)
g.add_edge(2, 0)
g.add_edge(2, 3)
g.add_edge(3, 3)

print("DFS Traversal:")
g.dfs()
```

# ***Iterative DFS (Stack)***

```python
def dfs_iterative(graph, start):
    visited = set()
    stack = [start]

    while stack:
        node = stack.pop()

        if node not in visited:
            print(node, end=" ")
            visited.add(node)

            # Add neighbors in reverse for correct order
            stack.extend(reversed(graph[node]))


# Driver Code
print("\nIterative DFS:")
dfs_iterative(graph, '5')
```

```python
from collections import deque

# Graph definition (REQUIRED)
graph = {
    "Arad": [("Zerind", 75), ("Sibiu", 140), ("Timisoara", 118)],
    "Zerind": [("Arad", 75), ("Oradea", 71)],
    "Oradea": [("Zerind", 71), ("Sibiu", 151)],
    "Sibiu": [("Arad", 140), ("Oradea", 151), ("Fagaras", 99), ("Rimnicu Vilcea", 80)],
    "Timisoara": [("Arad", 118), ("Lugoj", 111)],
    "Lugoj": [("Timisoara", 111), ("Mehadia", 70)],
    "Mehadia": [("Lugoj", 70), ("Drobeta", 75)],
    "Drobeta": [("Mehadia", 75), ("Craiova", 120)],
    "Craiova": [("Drobeta", 120), ("Rimnicu Vilcea", 146), ("Pitesti", 138)],
    "Rimnicu Vilcea": [("Sibiu", 80), ("Craiova", 146), ("Pitesti", 97)],
    "Fagaras": [("Sibiu", 99), ("Bucharest", 211)],
    "Pitesti": [("Rimnicu Vilcea", 97), ("Craiova", 138), ("Bucharest", 101)],
    "Bucharest": []
}


# BFS
def bfs(graph, start, goal):
    visited = set()
    queue = deque([(start, [start])])
    count = 0

    while queue:
        node, path = queue.popleft()
        count += 1

        if node == goal:
            return path, count

        if node not in visited:
            visited.add(node)

            for neighbor, _ in graph[node]:
                queue.append((neighbor, path + [neighbor]))

    return None, count


# DFS
def dfs_count(graph, start, goal):
    visited = set()
    stack = [(start, [start])]
    count = 0

    while stack:
        node, path = stack.pop()
        count += 1

        if node == goal:
            return path, count

        if node not in visited:
            visited.add(node)

            for neighbor, _ in graph[node]:
                stack.append((neighbor, path + [neighbor]))

    return None, count


# ✅ Correct place to call functions
bfs_path, bfs_count = bfs(graph, "Arad", "Bucharest")
dfs_path, dfs_count_val = dfs_count(graph, "Arad", "Bucharest")

print("BFS Path:", bfs_path)
print("BFS Nodes Visited:", bfs_count)

print("DFS Path:", dfs_path)
print("DFS Nodes Visited:", dfs_count_val)
```
