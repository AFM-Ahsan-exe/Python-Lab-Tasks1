# ***Uniformed Cost Search For Graph***

```python
import heapq   # Used for priority queue (min-heap)

# ---------------- UCS FUNCTION ----------------
def ucs(graph, start, goal):
    """
    Uniform Cost Search Algorithm

    graph  → dictionary (adjacency list)
    start  → initial node
    goal   → target node
    """

    # Priority Queue:
    # Stores (total_cost, current_node, path_so_far)
    # WHY? → So we always expand the lowest cost path first
    queue = [(0, start, [])]

    # Visited Dictionary:
    # Stores the minimum cost at which a node is reached
    # WHY? → Avoid revisiting nodes with higher cost (optimization)
    visited = {}

    # Loop runs until queue becomes empty
    while queue:

        # Pop element with smallest cost
        # heapq ensures lowest cost is always removed first
        cost, node, path = heapq.heappop(queue)

        # If we already visited this node with a lower cost → skip
        if node in visited and visited[node] <= cost:
            continue

        # Mark node as visited with current cost
        visited[node] = cost

        # Add current node to path
        path = path + [node]

        # Goal Test:
        # If current node is goal → return result
        if node == goal:
            return cost, path

        # Expand neighbors
        # graph[node] gives list of (neighbor, weight)
        for neighbor, weight in graph[node]:

            # Push new path into priority queue
            # New cost = old cost + edge weight
            heapq.heappush(queue, (cost + weight, neighbor, path))

    # If goal not found
    return float("inf"), []


# ---------------- GRAPH FROM IMAGE ----------------

# Adjacency list representation
# Format: node → [(neighbor, cost), ...]
graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('E', 3), ('D', 2)],
    'C': [('F', 5)],
    'D': [('G', 2)],
    'E': [('D', 1)],
    'F': [('G', 2)],
    'G': []  # Goal node has no outgoing edges
}


# ---------------- RUN UCS ----------------

start = 'A'   # Initial state
goal = 'G'    # Final state

cost, path = ucs(graph, start, goal)

# Print results
print("Shortest Path:", " -> ".join(path))
print("Total Cost:", cost)
```
